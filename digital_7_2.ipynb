## (7) 極と零点の配置

極：$p = \rho e^{\pm j\frac{\pi}{2}}$

零点：$z = e^{\pm j \frac{\pi}{4}}, e^{\pm j \frac{3\pi}{4}}$


\begin{align*}
分子 &= \left(z-e^{j\frac{\pi}{4}}\right)\left(z-e^{-j\frac{\pi}{4}}\right)
\left(z-e^{j\frac{3\pi}{4}}\right)\left(z-e^{-j\frac{3\pi}{4}}\right)\\
&= z^4 -2 \left( \cos \frac{3\pi}{4} + \cos \frac{\pi}{4} \right)z^3
+ 2\left( 1 + 2\cos \frac{\pi}{4} \cos \frac{3\pi}{4}\right)z^2
-2 \left( \cos \frac{3\pi}{4} + \cos \frac{\pi}{4} \right)z + 1\\
分母 &= \left( z-\rho e^{j\frac{\pi}{2}} \right)\left( z-\rho e^{-j\frac{\pi}{2}} \right)\\
&= z^2 -2\rho \cos \frac{\pi}{2} z + \rho^2\\
H(z) &=
\frac{z^4 -2 \left( \cos \frac{3\pi}{4} + \cos \frac{\pi}{4} \right)z^3
+ 2\left( 1 + 2\cos \frac{\pi}{4} \cos \frac{3\pi}{4}\right)z^2
-2 \left( \cos \frac{3\pi}{4} + \cos \frac{\pi}{4} \right)z + 1\\}{z^2 -2\rho \cos \frac{\pi}{2} z + \rho^2}\\
&= \frac{1 -2 \left( \cos \frac{3\pi}{4} + \cos \frac{\pi}{4} \right)z^{-1}
+ 2\left( 1 + 2\cos \frac{\pi}{4} \cos \frac{3\pi}{4}\right)z^{-2}
-2 \left( \cos \frac{3\pi}{4} + \cos \frac{\pi}{4} \right)z^{-3} + z^{-4}\\}
{z^{-2} -2\rho \cos \frac{\pi}{2} z^{-3} + \rho^2 z^{-4}}\\
\end{align*}

```python
omega = np.linspace(-np.pi, np.pi, 1024, endpoint=False)

rho = 0.7

t1 = -2.0 * np.cos(3.0 * np.pi / 4.0) - 2.0 * np.cos(np.pi / 4.0)
t2 = 2.0 + 4.0 * np.cos(np.pi / 4.0) * np.cos(3.0 * np.pi / 4.0)
b = [1.0, t1, t2, t1, 1.0]
a = [0.0, 0.0, 1.0, -2.0 * rho * np.cos(np.pi / 2.0), rho**2]

_, H = signal.freqz(b, a, omega)
absH = np.abs(H)
argH = np.angle(H)

fig1 = plt.figure()
ax1 = fig1.add_subplot(1, 1, 1)
ax1.plot(omega, absH)
ax1.grid()

fig2 = plt.figure()
ax2 = fig2.add_subplot(1, 1, 1)
ax2.plot(omega, argH)
ax2.grid()
```

### (a) 振幅特性

\begin{align*}
\left| H_1(e^{j\omega}) \right|^2 &= \frac{1}{1+ae^{-j\omega}}\frac{1}{1+ae^{j\omega}}\\
&= \frac{1}{1 + 2a \cos \omega +a^2}\\
\left| H_2(e^{j\omega}) \right|^2 &= \frac{1}{a+e^{-j\omega}}\frac{1}{a+e^{j\omega}}\\
&= \frac{1}{a^2 + 2a \cos \omega +1}
\end{align*}

上式から$\left| H_1(e^{j\omega}) \right|=\left| H_2(e^{j\omega}) \right|$
