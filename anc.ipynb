```python
import numpy as np
import unittest
```

```python
class ANC:
    def __init__(self, order=20, delay=20):
        self.firbuf = np.zeros(order)
        self.delaybuf = np.zeros(delay)
        return
    
    def update_buf(self, buf, new):
        updated = np.roll(buf, 1)
        updated[0] = new
        return updated
    
class TestANC(unittest.TestCase):
    def test_update_buf(self):
        anc = ANC()
        result = anc.update_buf(np.array([2, 3, 0]), 1)
        expected = np.array([1, 2, 3])
        self.assertTrue(np.alltrue(result==expected))


unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
