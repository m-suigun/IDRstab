```python
import numpy as np
from matplotlib import pyplot as plt
from scipy.signal import freqz
```

```python
fs = 125.0/2.0

omega = np.linspace(0, np.pi, 512)
H1 = np.sqrt(2.0 * (1.0 - np.cos(omega)))
omega2, H2 = freqz([1.0, -1.0])
plt.plot(2*fs-fs*omega/np.pi, H1)
plt.plot(2*fs-fs*omega2/np.pi, np.abs(H2))
plt.show()
```
