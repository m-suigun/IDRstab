## 2.3

(1)

\begin{align*}
E[x(n)^2] &= E[\{\alpha x(n-1) + u(n-1)\}\{\alpha x(n-1) + u(n-1)\}]\\
&= E[\alpha^2 x(n-1)^2 + u(n-1)^2]\\
&= \alpha^2E[x(n-1)^2]+1\\
\end{align*}

定常な過程なので，$E[x(n)^2]=E[x(n-1)^2]$．上の式に代入して整理すると以下が得られる．

\begin{align*}
E[x(n)^2] &= \frac{1}{1-\alpha^2}
\end{align*}


(2)

\begin{align*}
E[x(n)x(n+\ell)] &= E[x(n)(\alpha x(n+\ell-1) + u(n+\ell-1))]\\
&= \alpha E[x(n)x(n+\ell-1)]\\
&= \alpha E[x(n)(\alpha x(n+\ell-2) + u(n+\ell-2))]\\
&= \alpha^2 E[x(n)x(n+\ell-2)]\\
&\cdots \\
&= \alpha^\ell E[x(n)^2]\\
&= \frac{\alpha^\ell}{1-\alpha^2}
\end{align*}
